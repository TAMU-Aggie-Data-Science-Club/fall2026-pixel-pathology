# **2. Short explanation of the image_id, lesion_id, and dx columns**



### **1.  image_id column**
        * The image_id column consists of 10015 records with a datatype of object.
        * These records are unique identifiers for individual skin images.
        * There records are in the format beggining with "ISIC_" and ending with a 7 digit number.
        * It serves as the primary key of the database, connecting data entries directly to image file names.
### **2. "lesion_id column**
        * The lesion_id column has 10015 records with the datatype object.
        * These records are unique identifiers for specific types of skin lesions.
        * It appears that multiple images can belong to the same lesion type.
        * It is formatted as HAM_00 followed by 5 digits
        * One lesion_id can correspond to multiple image_id entries because single lesions were often photographed at different angles or time points

### **3. "dx column**
        * The dx column has 10015 records with the datatype object,
        * These records are the diagnosis category assigned to that particular skin lesion.
        * It has 7 distinct classes - nv: melanocytic nevi(benign moles),
          mel: melanoma(malignant skin cancer),
          bkl: benign keratosis-like lesions(solar lentigines, seborrheic keratoses, lichen-planus-like keratoses),
          bcc: basil cell carcenoma (malignant),
          akiec: Actinic keratoses/intraepithelial carcinoma (pre-cancerous/bowen's disease),
          vasc: vascular lesions (angiomas, cherry angiomas, pyogenic granulomas, hemorrhage),
          dfb: dermatofibroma (benign skin lesion)
        * As shown below, bkl means benign keratosis-like lesions diagnosis.
        * This is a column the model can learn to predict.

# **Further Exploration of Columns Below:**

```python
target_cols = ['image_id', 'lesion_id', 'dx']
summary_df = pd.DataFrame({
    'Data Type': df[target_cols].dtypes,
    'Total Count': df[target_cols].count(),
    'Unique Values': df[target_cols].nunique(),
    'Missing Values': df[target_cols].isnull().sum()
})

print("=== Target Columns Summary ===")
print(summary_df)
print("\n" + "="*40 + "\n")

# 4. Multi-image lesion analysis
lesion_counts = df['lesion_id'].value_counts()
multi_image_lesions = (lesion_counts > 1).sum()
max_images = lesion_counts.max()

print("=== Lesion vs Image Breakdown ===")
print(f"Total Images (image_id): {df['image_id'].nunique()}")
print(f"Total Unique Lesions (lesion_id): {df['lesion_id'].nunique()}")
print(f"Lesions with multiple images: {multi_image_lesions} ({(multi_image_lesions/df['lesion_id'].nunique())*100:.1f}%)")
print(f"Max images recorded for a single lesion: {max_images}")
print("\n" + "="*40 + "\n")

# 5. Diagnostic class distribution (dx)
dx_mapping = {
    'nv': 'Melanocytic nevi',
    'mel': 'Melanoma',
    'bkl': 'Benign keratosis-like lesions',
    'bcc': 'Basal cell carcinoma',
    'akiec': 'Actinic keratoses',
    'vasc': 'Vascular lesions',
    'df': 'Dermatofibroma'
}

dx_summary = df['dx'].value_counts().reset_index()
dx_summary.columns = ['dx_code', 'image_count']
dx_summary['diagnosis_label'] = dx_summary['dx_code'].map(dx_mapping)
dx_summary['percentage'] = (dx_summary['image_count'] / len(df)) * 100
```

# **3. A table or chart showing the number of images in each class.**

```python
# Displayed image counts and percentages for each diagnosis
class_table = dx_summary[
    ['dx_code', 'diagnosis_label', 'image_count', 'percentage']
].copy()

class_table['percentage'] = class_table['percentage'].round(2)
display(class_table)
```

### Class Count Explanation

This table summarizes the number of images in each of the seven diagnostic classes. Images were counted using the `dx` column, and each percentage was calculated by dividing the class count by the total of **10,015 images** and multiplying by 100. The classes are displayed from highest to lowest image count.

# **4. Three observations about the dataset.**

1. The dataset has 10,015 images, but the classes aren't evenly divided. The nv class has the most images, which could make it harder for the model to learn the smaller classes.

2. Some lesions have multiple images. Each image has a different image_id, but they can share the same lesion_id.

3. The dataset includes both benign and cancerous skin conditions. Some can look similar, so the model needs to learn how to tell them apart.

# **5. A short explanation of why images sharing a lesion_id must remain in the same train, validation, or test split.**
* Simpy put, as to make sure the model is able to generalize skin lesions across multiple patients. If the model is provided with the same patient/same legion at multiple different IDs, then it memorizes the things specific to that patient, like their skin texture, color or the kind of legion on them. If this is the case, then the model will be weaker at correctly identifying lesions on people that are unlike this one patient.
